## Main concepts
### 1. What is Queueing Theory?

Queueing theory is a branch of applied mathematics and operations research that studies waiting lines and congestion in systems where entities arrive to request a service from limited resources. It provides mathematical models for analyzing arrival patterns, waiting times, queue lengths, service rates, and resource utilization. The main objective is to understand how a service system performs under different demand and capacity conditions and to identify ways to reduce waiting times, improve efficiency, and allocate resources effectively.
Queueing theory is applicable to a wide range of sectors, including healthcare, public administration, transportation, telecommunications, and manufacturing. For example, in healthcare, it can be used to estimate patient waiting times and determine the appropriate number of medical staff. In public administration, it can help evaluate citizen queues at service counters and optimize staffing levels. In public transportation, it can support the analysis of passenger waiting times, vehicle capacity, and service frequency.
A queueing system is typically characterized by its arrival process, service mechanism, number of servers, queue capacity, and service discipline—the rule determining the order in which customers are served. Mathematical models such as M/M/1 and M/M/c allow researchers to estimate key performance indicators, including average waiting time, average queue length, server utilization, and the probability that an arriving customer must wait.
In the context of public services, queueing theory provides a quantitative framework for understanding the relationship between demand, service capacity, and service quality. Its findings can inform decisions about staffing, scheduling, resource allocation, and operational improvements (Kumar et al., 2009).

### 2. Why is Queueing Theory important in public administration?

Public institutions often face limited budgets, staff shortages, fluctuating demand, and legal deadlines. Queueing theory provides a framework for balancing service demand with available capacity (Kleinrock, 1975).

***Reducing waiting times***

Identifies bottlenecks and helps determine how many service counters or employees are needed.

***Improving citizen satisfaction***

Reduces long queues, uncertainty, and unnecessary visits to public offices.

***Optimizing public resources***

Helps allocate employees, service counters, and budgets according to demand.

***Supporting evidence-based decisions***

Enables administrators to compare different staffing levels and service-delivery models.

### 3. Characteristics

**1. The Calling Population**

The population of potential customers, referred to as the calling population, will be assumed to be infinite, even though the number of potential customers is actually finite. When the population of potential customers is large this assumption is innocuous and actually can simplify the model. This is especially true when we believe that at any given time the number of customers being served or waiting for service is a small proportion of the whole population (Leonelli, 2021).

The assumption of an infinite population is such that the rate of arrival of customers is not affected by the number of customers that have already joined the queuing system. In addition, this will usually entail that the rate of arrival is constant throughout time.

**2. System Capacity**

In many queuing systems there is a limit to the number of customers that may be in the waiting line or system. An arriving customer who finds the system full does not enter but returns immediately to the calling population. However, there are other systems that may simply have an infinite capacity. When a system has a limited capacity, a distinction is made between the arrival rate (i.e. the number of arrivals per time unit) and the effective arrival rate (the number who arrive and enter the system per time unit).

**3. The Arrival Process**

The arrival process for infinite population models is usually characterized in terms of inter-arrival times of successive customers. Arrivals may occur at scheduled times or random times. When at random times, the inter-arrival times are usually characterized by a probability distribution. In addition, customers may arrive one at a time or in batches. The batch may be of constant size or of random size.

The most important model for random arrivals is the Poisson arrival process (mathematical model used to describe random, independent events — such as customer arrivals — that occur at a constant average rate $λ$ over time). If $A_n$ represents the inter-arrival time between customer $n-1$ and customer $n$, then for a Poisson arrival process $A_n$ is exponentially distributed with mean $1/\lambda$ per time unite. The arrival rate is $\lambda$ customers per time unit. The number of arrivals in a time interval of length $t$ has the Poisson distribution with mean $\lambda t$ customers (Leonelli, 2021).

**4. Queue Behavior and Queue Discipline**

Queue behavior refers to the actions of customers while in a queue waiting for service to begin. In some situation, there is a possibility that incoming customers will balk, renege, or jockey (move from one line to another if they think they have chosen a slow line).

Queue discipline refers to the logical ordering of customers in a queue and determines which customer will be chosen for service when a server becomes free. Common queue disciplines include (Smartqueue, 2024): 

 - First-Come, First-Served (FCFS) / First-In, First-Out (FIFO): Customers are served in the exact order of their arrival. This is the most common and perceived as the fairest approach.

 - Last-Come, First-Served (LCFS) / Last-In, First-Out (LIFO): The most recently arrived customer is selected for service first (like an elevator or a stack of plates).

 - Service in Random Order (SIRO): Customers are chosen randomly regardless of their arrival time.

 - Priority-Based Discipline: Customers are divided into classes, and higher-priority tasks or customers are served before lower-priority ones. This can be preemptive (interrupting a lower-priority customer) or non-preemptive (waiting for the current customer to finish).

 - Round-Robin (RR): Each customer gets a small time slice of service in turns; if not finished, they move to the back of the line. This is common in computer operating systems and processor scheduling.

 - Shortest Job First (SJF) / Shortest Remaining Processing Time (SRPT): The customer or task with the shortest estimated service time is processed first.

**5. Service Times and Service Mechanism**

The service times of successive arrivals are denoted by $S_1,S_2,…$. They may be constant or of random duration. In the latter case $\{S_1,S_2,S_3,...\}$ is usually characterized as a sequence of independent and identically distributed random variables. Sometimes services are identically distributed for all customers of a given type or class or priority, whereas customers of different types might have completely different service-time distributions. In addition in some systems service times depend upon the time of the day or upon the length of the waiting line (Leonelli, 2021).

A queuing system consists of a number of service counters and interconnecting queues. Each service center consists of some number of server, $c$, working in parallel; that is, upon getting to the head of the line, a customer takes the first available server. Parallel service mechanisms are either single server $(c = 1)$, multiple server $(1 < c < ∞)$, or unlimited servers $(c = ∞)$.

### 4. Kendall's notation

Recognizing the diversity of queuing systems, a notational system was introduced in the 50’s which has been widely adopted. The convention is based on the format  $A/ B/ c/ N/ K$, where the letters represent the following system characteristics (Leonelli, 2021):

 - $A$ represents the inter-arrival time distribution

 - $B$ represents the service-time distribution

 - $c$ represents the number of parallel servers

 - $N$ represents the system capacity

 - $K$ represents the size of the calling population

Common symbols for $A$ and $B$ are include $M$ (exponential or Markov), $D$ (constant or deterministic) and $G$ (arbitrary or general).

For example, $M/M/1/∞/∞$ indicates a single-server system that has unlimited queue capacity and infinite population of potential arrivals. The inter-arrival times and service times are exponentially distributed. When $N$ and $K$ are infinity, they may be dropped from the notation. For example $M/M/1/∞/∞$ is often shorted to $M/M/1$.
All systems will be assumed to have a FIFO queue discipline.

### 5. Key concepts and measures of performance

Queueing systems are commonly analyzed using the following variables (Gross et al., 2008; Erlang, 1917):

| Symbol | Definition | Typical unit / interpretation |
|---|---|---|
| $\lambda$ | Arrival rate | Entities per minute or hour |
| $\mu$ | Service rate of **one** server | Entities served per minute or hour |
| $c$ | Number of parallel, identical servers | Count |
| $1/\lambda$ | Interarrival time, for a stationary Poisson arrival process | Time per arrival |
| $1/\mu$ | Service time of one server under exponential service | Time per service |
| $a=\lambda/\mu$ | Offered load | Erlangs; workload expressed in server-equivalents |
| $\rho=\lambda/(c\mu)$ | Utilization of the total server pool in an M/M/c model | Fraction, between 0 and 1 for a stable model |
| $L$ | Number of customers in the entire system (waiting + in service) | Entities |
| $L_q$ | Number of customers waiting in the queue, excluding those in service | Entities |
| $W$ | Average time in the system (waiting + service) | Time |
| $W_q$ | Average waiting time before service begins | Time |
| $P_0$ | Steady-state probability that the system is empty | Probability |
| $P\mathrm{wait}$ | Probability an arrival must wait before starting service in M/M/c (Erlang C) | Probability |


For a stable, single-server queue, the utilization rate is: $\rho=\frac{\lambda}{\mu}$

For the basic single-server model, stability requires $\lambda < \mu$.

Leonneli suggests several examples for these performance measures to be estimated from simulation runs:

**1. Average Number of Customers in the System $L$**

Consider a queuing system over a period of time $T$ and let $L(t)$ denote the number of customers in the system at time $t$. Let $T_i$ be the total time in $[0,T]$ in which the system contained exactly $i$ customers. In general $\displaystyle\sum_{i=0}^{\infty}T_i = T$. The *average number of customers in the system* is estimated by 

$$\widehat{L} = \frac{1}{T}\sum_{i = 0}^{\infty}i T_i = \sum_{i = 0}^{\infty}i\frac{T_i}{T}$$

It could be noticed that $T_i/T$ is the proportion of time the system contains exactly $i$ customers.

Many queuing systems exhibit some kind of long-run stability in terms of their average performance. For such systems, as time $T$ gets large, the observed average number of customers in the system $\widehat{L}$ approaches a limiting value, say $L$, which is called the *long-run average number in system*. With probability 1 we have that

$$\widehat{L} = \frac{1}{T}\int_{0}^{T}L(t)dt\to L\text{ } as\text{ } T \to \infty$$

If a simulation run length $T$ is sufficiently long, the estimator $\widehat{L}$ becomes arbitrarily close to $L$.

The above equations can be applied to any subsystem of a queuing system. If $L_q(t)$ denotes the number of customers waiting in queue, and $\mathrm{T}_{i}^{q}$ denotes the total time in $[0,T]$ in which exactly $i$ customers are waiting in queue, then

$$\widehat{L}_q = \frac{1}{T}\sum_{i = 0}^{\infty}i\mathrm{T}_{i}^{q} = \int_{0}^{T}L_q(t)dt\to L_q \text{ } as \text{ } T \to \infty $$

**2. Average Time Spent in System per Customer $w$**

If we simulate a queuing system for some period of time, say $T$, then we can record the time that each customer spends in the system during $[0,T]$, say $W1,W2,…,W_N$ where $N$ is the number of arrivals in $[0,T]$. The average time spent in system per customer, called the *average system time*, is given by

$$\widehat{w} = \frac{1}{N}\sum_{i = 1}^{N}W_i$$

For stable systems, as $N \to \infty$ 

$$\widehat{w} \to w$$

with probability 1, where $w$ is called the *long-run average system time*.

If the system under consideration is the queue alone, we can re-write the above equations as

$$\widehat{w}_q = \frac{1}{N}\sum_{i = 1}^{N}\mathrm{W}_{i}^{q}\to w_q\text{ } as\text{ } N\to \infty$$

where $\mathrm{W}_{i}^{q}$ is the total time customer $i$ spends waiting in queue, $\widehat{w}_q$ is the observed average time spend in queue (or delay) and $w_q$ is the long-run average delay per customer.

**3. Little’s Law**

Little's Law relates the average queue size and waiting time: $L_q=\lambda W_q$ (Little, 1961)

This means that, over the long run, the average number of people waiting equals the arrival rate multiplied by the average time spent waiting, provided the system meets the law's steady-state assumptions.


### 6. Practical examples

***ILLUSTRATIVE SCENARIO I*** 

Let's have a municipal office that processes applications for official documents.


| Citizen arrivals |  |  | Service capacity |
|:--------:|:--------:|:--------:|:--------:|
|  24/hour   |     |     |   10/hour per employee  |


The office initially has two employees serving citizens.
- Total service capacity: 20 citizens per hour.
- Arrival rate: 24 citizens per hour.
- Utilization: \(24/20=120\%\).

Problem: Citizens arrive faster than the two employees can serve them. The queue will tend to grow rather than reach a stable long-run average.

Potential solution: a third employee could be added during peak hours, the application process should be improvev, online appointments coulg be introduced, or simple requests could be redirected to digital self-service (Hopp, Spearman, 2011).

With three employees, the nominal service capacity becomes 30 citizens per hour, and utilization falls to \(24/30=80\%\). This makes a stable queue possible under the basic model, although actual waiting times also depend on arrival patterns, service-time variability, and queue organization.

***ILLUSTRATIVE SCENARIO II*** 

An $M/M/1$ queueing model in a hospital setting represents a single-server station (such as a single triage nurse, a registration desk, or a specialty clinic doctor) where patient arrivals follow a Poisson distribution and service times follow an exponential distribution (Joseph, 2020).

 - Setting: A hospital specialty registration or single-doctor intake clinic.

 - Arrival Rate $(\lambda)$: An average of 15 patients per hour arrive randomly.

 - Service Rate $(\mu)$: The doctor/station can treat an average of 20 patients per hour (meaning it takes an average of 3 minutes per patient).

1. Traffic Intensity / Server Utilization $(\rho)$: $\rho =\frac{\lambda }{\mu }=\frac{15}{20}=0.75\text{\ (or\ }75\%\text{\ busy)}$

2. Average Number of Patients in the System $(L)$: $L=\frac{\rho }{1-\rho }=\frac{0.75}{1-0.75}=3\text{\ patients}$

3. Average Time a Patient Spends in the System $(W)$: $W=\frac{L}{\lambda }=\frac{3}{15}=0.2\text{\ hours\ }(\text{or\ }12\text{\ minutes})$

4. Average Number of Patients Waiting in Line $(L_q)$: $L_{q}=\frac{\rho ^{2}}{1-\rho }=\frac{0.75^{2}}{0.25}=2.25\text{\ patients}$

5. Average Waiting Time in the Queue $(W_q)$: $W_{q}=\frac{L_{q}}{\lambda }=\frac{2.25}{15}=0.15\text{\ hours\ }(\text{or\ }9\text{\ minutes})$

*Key takeaways:*

• Utilization $(\rho = 0.75)$: The doctor is busy 75% of the time, leaving 25% idle time to handle random surges.

• Bottlenecks: If the arrival rate $(\lambda)$ climbs closer to the service rate $(\mu = 20)$, utilization approaches 1, and wait times $(W_q)$ grow exponentially, signaling a need for an extra server (moving to an $M/M/c$ or multi-server model).

### 7. Types of queueing models used in government services

| Model | Description | Application |
|---|---|---|
| M/M/1 | Random arrivals, exponential service times, one server | A small public service desk |
| M/M/3 | Random arrivals, exponential service, three parallel servers | Three staffed counters sharing one line |
| M/M/c | Random arrivals, exponential service times, multiple servers | A municipal office with several counters |

These models simplify reality. Administrators should select a model that reflects actual arrival patterns, service times, priority rules, and operational constraints (Hopp, Spearman, 2011).

### 8. How public administrators can apply queueing theory
1. Collecting data. Recording arrival times, processing times, queue lengths, staffing levels, and peak periods.
2. Measuring performance. Calculating average waiting time, average queue length, service throughput, and employee utilization.
3. Identifing bottlenecks. Determining whether delays result from insufficient staff, complicated procedures, document verification, or IT systems.
4. Comparing alternatives. Evaluating additional employees, separating queues for different services, online applications, appointment systems, and simplified procedures.
5. Monitoring outcomes. Comparing waiting times, costs, citizen satisfaction, and equitable access before and after changes.

### 1. Queueing concepts and assumptions

Let $\lambda$ be the mean arrival rate, $\mu$ the mean service rate per server, and $c$ the number of parallel servers. Offered load is $\alpha=\lambda/\mu$, and $M/M/c$ utilization is $\rho=\lambda/(c\mu)$. The steady-state $M/M/c$ model requires $\rho<1$.

For an M/M/1 queue:

$ \rho=\lambda/\mu,\quad $
$ L=\frac{\rho}{1-\rho}, \quad$ 
$ L_q=\frac{\rho^2}{1-\rho},\quad $
$ W=\frac{1}{\mu-\lambda}, \quad $
$ W_q=\frac{\lambda}{\mu(\mu-\lambda)},\quad $ 
$ L=\lambda W,\quad L_q=\lambda W_q $

For an M/M/c queue:

$ P_0=\left[\sum_{n=0}^{c-1}\frac{\alpha^n}{n!}+\frac{\alpha^c}{c!(1-\rho)}\right]^{-1} $

$ P(\text{wait})=\frac{\alpha^c}{c!(1-\rho)}P_0,\quad $
$ W_q=\frac{P(\text{wait})}{c\mu-\lambda}, \quad $
$ L_q=\lambda W_q,\quad W=W_q+1/\mu,\quad L=\lambda W $

Assumptions include Poisson arrivals, exponential service times, pooled first-come-first-served queue, steady-state demand, and no abandonment. Real systems may need time-varying rates, priorities, multiple stages, non-exponential service, or finite capacity.

### 3. Analytical M/M/c model

Rates use the same time unit. The function returns infinite queueing measures when the system is unstable (λ ≥ cμ).

```python
def mmc_metrics(lam, mu, c):
    """Steady-state M/M/c metrics; rates must use the same time unit."""
    if lam < 0 or mu <= 0 or int(c) != c or c < 1:
        raise ValueError("Require lam >= 0, mu > 0, and integer c >= 1.")
    c = int(c)
    a = lam / mu
    rho = lam / (c * mu)
    if rho >= 1:
        return {"lambda": lam, "mu": mu, "servers": c, "utilization": rho,
                "p_wait": np.nan, "Lq": np.inf, "Wq_min": np.inf,
                "W_min": np.inf, "L": np.inf, "stable": False}
    tail = a**c / (math.factorial(c) * (1-rho))
    p0 = 1 / (sum(a**n / math.factorial(n) for n in range(c)) + tail)
    p_wait = tail * p0
    Wq = p_wait / (c*mu-lam)
    Lq = lam * Wq
    W = Wq + 1/mu
    return {"lambda": lam, "mu": mu, "servers": c, "utilization": rho,
            "p_wait": p_wait, "Lq": Lq, "Wq_min": Wq*60,
            "W_min": W*60, "L": lam*W, "stable": True}

example = mmc_metrics(24, 10, 3)
display(pd.DataFrame([example]))
assert abs(example["Lq"] - 24 * example["Wq_min"]/60) < 1e-8
```

### 5. Sensitivity analysis

High utilization can make waiting times rise sharply as demand approaches capacity. This section tests demand from 80% to 130% of baseline, for both baseline and intervention staffing (Elalouf, Wachtel, 2022; Kumar et al., 2024).

```python
sensitivity_rows = []
for domain, p in systems.items():
    for mult in [0.8, 1.0, 1.1, 1.2, 1.3]:
        for c in [p["base"], p["intervention"]]:
            m = mmc_metrics(p["lambda"]*mult, p["mu"], c)
            sensitivity_rows.append({"Domain": domain, "Demand multiplier": mult,
                "Servers": c, "Utilization": m["utilization"],
                "Mean queue wait (min)": m["Wq_min"], "Stable": m["stable"]})
sensitivity = pd.DataFrame(sensitivity_rows)
display(sensitivity)
for domain in systems:
    fig, ax = plt.subplots(figsize=(7,4))
    for c, g in sensitivity[sensitivity.Domain == domain].groupby("Servers"):
        ax.plot(g["Demand multiplier"], g["Mean queue wait (min)"],
                marker="o", label=f"{c} servers")
    ax.set_title(f"Demand sensitivity — {domain}")
    ax.set_xlabel("Arrival-rate multiplier")
    ax.set_ylabel("Mean queue wait (minutes)")
    ax.legend()
    ax.grid(True, alpha=.3)
    plt.tight_layout()
    plt.show()
```

### 6. Discrete-event simulation: healthcare and public administration

The simulation uses Poisson arrivals and exponential service times, a pooled queue, multiple parallel servers, and a warm-up period. It produces individual waits, medians, and percentiles as well as means (Joseph, 2020; Kumar et al., 2024). 

```python
def simulate_mmc(lam, mu, c, horizon_hours=240, warmup_hours=24, seed=1234):
    """Simulate M/M/c using SimPy; return completed-customer metrics."""
    local_rng = np.random.default_rng(seed)
    env = simpy.Environment()
    resource = simpy.Resource(env, capacity=c)
    records = []

    def customer(cid, arrival):
        with resource.request() as req:
            yield req
            start = env.now
            yield env.timeout(local_rng.exponential(1/mu))
            departure = env.now
            if arrival >= warmup_hours and departure <= horizon_hours:
                records.append({"id": cid, "arrival": arrival, "start": start,
                    "departure": departure, "wait": start-arrival,
                    "system_time": departure-arrival})

    def source():
        cid = 0
        while True:
            yield env.timeout(local_rng.exponential(1/lam))
            if env.now > horizon_hours:
                break
            cid += 1
            env.process(customer(cid, env.now))

    env.process(source())
    env.run(until=horizon_hours)
    return pd.DataFrame(records)

sim_rows = []
for idx, row in analytical.iterrows():
    sim = simulate_mmc(row["lambda"], row["mu"], int(row["servers"]), seed=SEED+idx)
    sim_rows.append({"Domain": row["Domain"], "Scenario": row["Scenario"],
        "Completed": len(sim),
        "Simulated mean wait (min)": sim["wait"].mean()*60 if len(sim) else np.nan,
        "Median wait (min)": sim["wait"].median()*60 if len(sim) else np.nan,
        "90th percentile wait (min)": sim["wait"].quantile(.90)*60 if len(sim) else np.nan,
        "Analytical mean wait (min)": row["Wq_min"]})
simulation_comparison = pd.DataFrame(sim_rows)
display(simulation_comparison)
```

### 7. Public transport: periodic batch-service simulation

Buses do not behave like continuously available counters. Passengers arrive over time, then a bus boards a finite batch at scheduled headways. If demand exceeds the available spaces, passengers remain for a later bus.

A simple random-arrival approximation gives expected wait near \(h/2\) for headway \(h\), **only if capacity is sufficient and arrivals are independent of bus times**. Capacity constraints can create residual queues and much longer waits (Wang et al., 2021).

```python
def simulate_bus_stop(lam_per_hour=90, headway_min=10, bus_capacity=25,
                      horizon_hours=8, seed=2026):
    """Poisson passenger arrivals; buses provide periodic finite batch service."""
    rng = np.random.default_rng(seed)
    arrivals, t = [], 0.0
    horizon_min = horizon_hours*60
    while True:
        t += rng.exponential(60/lam_per_hour)
        if t > horizon_min:
            break
        arrivals.append(t)
    arrivals = np.array(arrivals)
    waiting, i, waits, q_after = [], 0, [], []
    bus_times = np.arange(headway_min, horizon_min+headway_min, headway_min)
    for bus_t in bus_times:
        while i < len(arrivals) and arrivals[i] <= bus_t:
            waiting.append(arrivals[i]); i += 1
        for _ in range(min(bus_capacity, len(waiting))):
            waits.append(bus_t-waiting.pop(0))
        q_after.append(len(waiting))
    return {"arrivals": len(arrivals), "boarded": len(waits),
        "left_waiting_at_end": len(waiting)+(len(arrivals)-i),
        "mean_wait_min": float(np.mean(waits)) if waits else np.nan,
        "median_wait_min": float(np.median(waits)) if waits else np.nan,
        "p90_wait_min": float(np.quantile(waits,.90)) if waits else np.nan,
        "max_queue_after_bus": max(q_after, default=0),
        "bus_times": bus_times, "queue_after_bus": np.array(q_after),
        "waits": np.array(waits)}

transit_scenarios = [
    ("Baseline: 10 min, capacity 25", 10, 25),
    ("More frequent: 7.5 min, capacity 25", 7.5, 25),
    ("Higher capacity: 10 min, capacity 35", 10, 35),
]
transit_results, transit_details = [], {}
for j, (name, headway, cap) in enumerate(transit_scenarios):
    r = simulate_bus_stop(90, headway, cap, seed=SEED+j)
    transit_details[name] = r
    transit_results.append({"Scenario": name, "Arrivals": r["arrivals"],
        "Boarded": r["boarded"], "Still waiting at end": r["left_waiting_at_end"],
        "Mean wait (min)": r["mean_wait_min"], "Median wait (min)": r["median_wait_min"],
        "90th percentile wait (min)": r["p90_wait_min"],
        "Max queue after bus": r["max_queue_after_bus"],
        "Nominal capacity/hour": 60/headway*cap})
transit_table = pd.DataFrame(transit_results)
display(transit_table)

fig, ax = plt.subplots(figsize=(9,4))
for name, r in transit_details.items():
    ax.plot(r["bus_times"], r["queue_after_bus"], marker=".", label=name)
ax.set_title("Passengers left waiting after each bus")
ax.set_xlabel("Time (minutes)")
ax.set_ylabel("Queue after bus departure")
ax.legend()
ax.grid(True, alpha=.3)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(8,4))
for name, r in transit_details.items():
    if len(r["waits"]):
        ax.hist(r["waits"], bins=25, alpha=.4, label=name)
ax.set_title("Passenger waiting-time distributions")
ax.set_xlabel("Wait to boarding (minutes)")
ax.set_ylabel("Passengers")
ax.legend()
plt.tight_layout()
plt.show()
```

### Transit capacity condition

A necessary (not sufficient) long-run condition against queue accumulation is: $\lambda < \frac{60}{h}K$
where $h$ is headway in minutes and $K$ is the passenger capacity available for boarding at the stop. This ignores variable usable capacity, alighting, traffic delays, and route interactions (Vuchic, 2005).

At 90 passengers/hour:
- 10-minute headway and capacity 25: nominal capacity 150 passengers/hour.
- 7.5-minute headway and capacity 25: nominal capacity 200 passengers/hour.
- 10-minute headway and capacity 35: nominal capacity 210 passengers/hour.

Even when average capacity exceeds demand, random surges can create temporary queues.

## Optimization case study: Optimizing Waiting Time in Public Administration

***This case study is developed on the basis of core concepts and mathematical formulations articulated in Dr. Sztrik’s book.***

### 1. Queueing concepts and assumptions

The baseline model is an $M/M/c$ queue: citizens arrive according to a Poisson process, service times are exponentially distributed, and $c$ parallel employees serve citizens from a shared queue.

**1.1. Model parameters**

| Symbol | Definition | Unit |
|---|---|---|
| $\lambda_t$ | Citizen arrival rate during interval $t$ | Citizens/hour |
| $\mu_t$ | Service rate per employee during interval $t$ | Citizens/hour/employee |
| $c_t$ | Number of active service counters during interval $t$ | Counters |
| $\rho_t$ | Server utilization | Proportion |
| $P_0,_t$ | Probability that the system is empty | Probability |
| $P\mathrm{wait},t$ | Probability that an arriving citizen must wait | Probability |
| $L_q,_t$ | Expected number of citizens waiting | Citizens |
| $W_q,_t$ | Expected waiting time before service | Hours |
| $W_t$ | Expected total time, including service | Hours |
| $C_t$ | Cost of operating one counter during interval $t$| Currency/hour |

The decision variable is $c_t$, the number of counters to open in each interval. The arrival and service rates are estimated from operational data.

**1.2. Server utilization and stability**

The utilization of the service capacity is $\rho_t = \frac{\lambda_t}{c_t\mu_t}$

For a stable $M/M/c$ queue, the required condition is $\lambda_t\lt c_t\mu_t$ or $\rho_t\lt 1$

If utilization approaches 1, waiting times can rise sharply. A system with utilization above or equal to 1 has no finite steady-state mean waiting time under the standard model.

**1.3. Probability that the system is empty**

Define the offered load: $\alpha_t = \frac{\lambda_t}{\mu_t}$

Then the probability that no citizen is waiting or being served is 
$$P_0,_t = \left[ \sum_{n = 0}^{c_t-1}\frac{\mathrm{\alpha}_{t}^{n}}{n!} + \frac{\mathrm{\alpha}_{t}^{c_t}}{c_t!(1-\rho_t)} \right]^{-1}$$

**1.4. Probability that a citizen must wait: Erlang C**

The probability that all counters are occupied when a citizen arrives is
$$P\mathrm{wait},t = \frac{\mathrm{\alpha}_{t}^{c_t}}{c_t!(1-\rho_t)}P_0,_t$$

**1.5. Expected waiting time and queue length**

The expected time spent waiting before service begins is $W_q,_t = \frac{P\mathrm{wait},t}{c_t\mu_t - \lambda_t}$

The expected queue length follows from Little’s Law: $L_q,_t = \lambda_tW_q,_t$

The expected total time in the system is $W_t = W_q,_t + \frac{1}{\mu_t}$

and the expected number of citizens in the whole system is $L_t = \lambda_tW_t$

### 2. Mathematical optimization of staffing levels

Queueing theory estimates how the system performs for a given number of counters. Optimization determines the number of counters that should be open.
Divide the working day into $T$ time intervals, for example, 30-minute intervals. The optimization model selects a staffing level $c_t$ for every interval.

**2.1. Decision variables**

Let $c_t\in Z_\ge 0$ and $t = 1,....,T$ where  $c_t$ is the number of counters operating during interval $t$.

Let $\Delta t$ be the duration of each interval in hours. The expected number of arrivals in that interval is $A_t = \lambda_t\Delta t$

**2.2. Objective function A: minimize average waiting time**

The first objective is to minimize the arrival-weighted mean waiting time:

$$\overset{min}{\left\{ c_t \right\}}\frac{\displaystyle\sum_{t = 1}^{T}\lambda_t\Delta tW_q(\lambda_t,\mu_t,c_t)}{\displaystyle\sum_{t = 1}^{T}\lambda_t\Delta_t}$$

This objective gives more weight to intervals in which more citizens arrive.

However, minimizing waiting time alone may lead to excessive staffing. Therefore, the model needs practical constraints on staffing availability, cost, and service quality.

**2.3. Objective function B: minimize staffing cost**

An alternative is to minimize the cost of staffing the counters:

$$\min_{\{c_t\}} \sum_{t=1}^{T} C_t c_t\Delta t$$

where $C_t$ is the cost of operating one counter per hour.

The optimization must satisfy a waiting-time or service-level requirement. This is often a more practical formulation for public administration because it seeks the lowest staffing cost that achieves a defined service standard.

**2.4. Constraints**

*Capacity constraint:* $1\leq c_t\leq c_{\max,t}$

Here, $c_{\max,t}$ is the maximum number of counters that can be staffed during interval $t$. If the office can close entirely during some intervals, the lower bound can be adjusted.

*Queue stability constraint:* $\lambda_t<c_t\mu_t$

This prevents the selected staffing level from having insufficient service capacity under the $M/M/c$ model.

*Budget constraint:* $\displaystyle\sum_{t=1}^{T}C_tc_t\Delta t\leq B$

where $B$ is the available staffing budget.

*Service-quality constraint:* 

The mean waiting time can be limited by a specified target: $W_{q,t}\leq W_{\max}$

For example, $W_{\max}=10$ minutes. Alternatively, the model can use a probabilistic service-level target, which is often more informative.

### 3. Service-level optimization

A mean waiting-time target does not describe how waiting times are distributed. A service-level requirement can specify the proportion of citizens who must begin service within a defined period.
For a stable $M/M/c$ queue, the probability that queue waiting time is no greater than $\tau$ is

$$P(W_q\leq\tau)=1-P_{\mathrm{wait}}e^{-(c\mu-\lambda)\tau}$$

where $\tau$ is expressed in the same time units as the arrival and service rates.

For example, the office might require at least 80% of citizens to begin service within 10 minutes: $P(W_{q,t}\leq 10/60)\geq 0.80$

This example assumes rates are measured per hour. The target is illustrative, not an official standard.

### 4. Discrete-event simulation for validation

The final stage is to validate candidate staffing schedules with a simulation. This is important because the actual queue evolves continuously throughout the day.
In a simulation, each citizen has an arrival time $\alpha_i$, a service-start time $s_i$, and a departure time $d_i$.

Individual waiting time: $w_i=s_i-\alpha_i$

Individual total time in the system: $v_i=d_i-\alpha_i$

For $N$ citizens served, the simulated mean waiting time is
$$\overline{W}_q=\frac{1}{N}\sum_{i=1}^{N}(s_i-\alpha_i)$$

The proportion starting service within the target time $\tau$ is
$$\widehat{SL}(\tau)=\frac{1}{N}\sum_{i=1}^{N}\mathbf{1}(w_i\leq\tau)$$

where $\mathbf{1}(\cdot)$ equals 1 when the condition is true and 0 otherwise.

The simulation can represent changing arrivals, queue carryover between intervals, staff shifts, breaks, and service-time variability. Candidate schedules can then be compared using average waiting time, the 90th percentile of waiting time, service level, and total staffing cost.


### 5. Optimizing Waiting Time Model
This baseline model demonstrates the number of service counters by time interval. It uses the M/M/c queue, Erlang C formulas, and a simple staffing search. All example inputs are hypothetical.

**Workflow:** arrival/service data → queue performance → minimum staffing level meeting service targets → plan → sensitivity analysis.

The analytical model treats each interval as approximately steady-state. 

```python
pd.set_option("display.max_columns", 30)
```

**5.1. Input data and policy targets**

Each row represents one time interval. Rates are per hour. `service_rate` is the number of citizens one employee can serve per hour. Targets below are examples.

```python
intervals = pd.DataFrame({
    "interval": ["09:00–10:00", "10:00–11:00", "11:00–12:00", "12:00–13:00"],
    "arrival_rate": [30, 40, 20, 15],
    "service_rate": [10, 10, 10, 10],
    "max_counters": [6, 6, 6, 6],
    "cost_per_counter_hour": [1.0, 1.0, 1.0, 1.0]
})

MAX_MEAN_WAIT_MIN = 10
SERVICE_LEVEL_MIN = 0.80
SERVICE_LEVEL_THRESHOLD_MIN = 10
intervals
```

**5.2. $M/M/c$ formulas**

For arrival rate $\lambda$, service rate per server $\mu$, and $c$ parallel servers, utilization is $\rho = \lambda/(c\mu)$. A stable queue requires $\lambda\lt c\mu$. The probability of waiting is given by Erlang C. The formulas below return mean queue waiting time, total time in the system, and the probability of starting service within a specified threshold.

```python
def mmc_metrics(lam, mu, c):
    """M/M/c steady-state metrics. λ and μ are rates per hour."""
    lam, mu, c = float(lam), float(mu), int(c)
    if lam < 0 or mu <= 0 or c < 1:
        raise ValueError("Require λ >= 0, μ > 0, and c >= 1.")
    rho = lam / (c * mu)
    if lam == 0:
        return dict(stable=True, utilization=0.0, P0=1.0, P_wait=0.0,
                    Lq=0.0, Wq_hours=0.0, Wq_minutes=0.0,
                    W_minutes=60/mu)
    if rho >= 1:
        return dict(stable=False, utilization=rho, P0=np.nan, P_wait=np.nan,
                    Lq=np.inf, Wq_hours=np.inf, Wq_minutes=np.inf,
                    W_minutes=np.inf)
    a = lam / mu
    p0_denominator = sum(a**n / math.factorial(n) for n in range(c))
    p0_denominator += a**c / (math.factorial(c) * (1-rho))
    P0 = 1 / p0_denominator
    P_wait = (a**c / (math.factorial(c) * (1-rho))) * P0
    Wq_hours = P_wait / (c*mu - lam)
    Lq = lam * Wq_hours
    W_hours = Wq_hours + 1/mu
    return dict(stable=True, utilization=rho, P0=P0, P_wait=P_wait, Lq=Lq,
                Wq_hours=Wq_hours, Wq_minutes=60*Wq_hours,
                W_minutes=60*W_hours)

def prob_start_within(lam, mu, c, threshold_minutes):
    """Probability queue wait is <= threshold; assumes stable M/M/c."""
    m = mmc_metrics(lam, mu, c)
    if not m["stable"]:
        return 0.0
    if m["P_wait"] == 0:
        return 1.0
    tau_hours = threshold_minutes / 60
    return 1 - m["P_wait"] * np.exp(-(c*mu-lam)*tau_hours)

# Example check: three counters, 24 arrivals/hour, 10 services/hour/counter.
pd.Series(mmc_metrics(24, 10, 3))
```

**5.3. Staffing optimization**

For each interval should be tested feasible counter counts and selected the smallest number meeting both targets:
- mean queue wait ≤ 10 minutes;
- at least 80% of arrivals begin service within 10 minutes.

If no available staffing level meets both targets, the result is flagged. This is an interval-by-interval optimization.

```python
def choose_staffing(row):
    candidates = []
    for c in range(1, int(row.max_counters) + 1):
        m = mmc_metrics(row.arrival_rate, row.service_rate, c)
        service_level = prob_start_within(
            row.arrival_rate, row.service_rate, c, SERVICE_LEVEL_THRESHOLD_MIN
        )
        meets = (m["stable"] and
                 m["Wq_minutes"] <= MAX_MEAN_WAIT_MIN and
                 service_level >= SERVICE_LEVEL_MIN)
        candidates.append({**m, "counters": c,
                           "service_level": service_level, "meets_targets": meets})
    feasible = [x for x in candidates if x["meets_targets"]]
    if feasible:
        return min(feasible, key=lambda x: x["counters"])
    best = max(candidates, key=lambda x: (x["service_level"], -x["Wq_minutes"]))
    best["meets_targets"] = False
    return best

rows = []
for _, row in intervals.iterrows():
    m = choose_staffing(row)
    rows.append({
        "interval": row["interval"],
        "arrivals_per_hour": row["arrival_rate"],
        "recommended_counters": m["counters"],
        "utilization": m["utilization"],
        "mean_queue_wait_min": m["Wq_minutes"],
        "mean_total_time_min": m["W_minutes"],
        "prob_start_within_10_min": m["service_level"],
        "targets_met": m["meets_targets"]
    })

staffing_plan = pd.DataFrame(rows)
staffing_plan.round(3)
```

**5.4. Visualization of the proposed staffing plan**

```python
fig, ax = plt.subplots()
ax.bar(staffing_plan["interval"], staffing_plan["recommended_counters"])
ax.set(title="Recommended counters by interval", xlabel="Time interval", ylabel="Counters")
ax.tick_params(axis="x", rotation=20)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots()
ax.plot(staffing_plan["interval"], staffing_plan["mean_queue_wait_min"], marker="o")
ax.axhline(MAX_MEAN_WAIT_MIN, linestyle="--", label="Mean waiting-time target")
ax.set(title="Predicted mean queue waiting time", xlabel="Time interval", ylabel="Minutes")
ax.tick_params(axis="x", rotation=20)
ax.legend()
plt.tight_layout()
plt.show()
```

**5.5. Comparison with a fixed staffing baseline**

The example compares the selected schedule with four counters open throughout the four-hour period. Costs are expressed in relative counter-hours.

```python
intervals["recommended_counters"] = staffing_plan["recommended_counters"]
intervals["optimized_cost"] = intervals["recommended_counters"] * intervals["cost_per_counter_hour"]
BASELINE_COUNTERS = 4
intervals["baseline_cost"] = BASELINE_COUNTERS * intervals["cost_per_counter_hour"]

comparison = pd.DataFrame({
    "schedule": ["Fixed baseline", "Optimized by interval"],
    "counter_hours": [
        BASELINE_COUNTERS * len(intervals),
        intervals["recommended_counters"].sum()
    ],
    "relative_cost": [intervals["baseline_cost"].sum(), intervals["optimized_cost"].sum()]
})
comparison
```

**5.6. Sensitivity analysis**

Tests on how the staffing recommendation changes if demand rises by 10%, 20%, or 30% should be conducted. This is useful when arrival rates are uncertain or seasonal.

```python
sensitivity_rows = []
for increase in [0.0, 0.10, 0.20, 0.30]:
    total_counters = 0
    all_targets_met = True
    for _, row in intervals.iterrows():
        stressed = row.copy()
        stressed["arrival_rate"] = row["arrival_rate"] * (1 + increase)
        m = choose_staffing(stressed)
        total_counters += m["counters"]
        all_targets_met = all_targets_met and m["meets_targets"]
    sensitivity_rows.append({
        "demand_increase_pct": int(increase*100),
        "total_counter_hours": total_counters,
        "all_intervals_meet_targets": all_targets_met
    })

pd.DataFrame(sensitivity_rows)
```

## Optimization case study: M/M/c queueing system with staffing optimization in hospitals

***This case study is developed on the basis of core concepts and mathematical formulations articulated in Dr. Sztrik’s book.***

This model focuses on one hospital service unit, such as an outpatient clinic or a hospital registration and examination unit.
The objective is to be determined the optimal number of doctors or service stations needed to minimize patient waiting time while keeping staffing costs under control.

### 1. Model description
The model assumes:
- Patients arrive randomly at the hospital according to a Poisson process.
- Patient service times follow an exponential distribution.
- There are $c$ doctors or service stations working in parallel.
- Patients wait in one common queue.
- Patients are served according to the First-Come, First-Served (FCFS) principle.
- Each doctor provides service at the same average rate.

### 2. Mathematical concepts and formulas
**2.1. Arrival rate $(\lambda)$**

The arrival rate represents the average number of patients arriving per hour.

For example, if 24 patients arrive during one hour:
$\lambda=24 \text{ patients/hour}$

**2.2. Service rate $(\mu)$**

The service rate represents the average number of patients examined by one doctor per hour.

If one consultation takes an average of 15 minutes: $\mu=\frac{60}{15}=4$

Therefore, one doctor can examine an average of 4 patients per hour.

**2.3. Number of doctors $(c)$**

The number of doctors working simultaneously is the decision variable in the optimization model.

For example: $c\in\{1,2,3,4,5,6\}$

**2.4. Probability of an empty system $(P_0)$**

For an $M/M/c$ queue:
$$P_0=\left[\sum_{n=0}^{c-1}\frac{(\lambda/\mu)^n}{n!}+\frac{(\lambda/\mu)^c}{c!(1-\rho)}\right]^{-1}$$

This is the probability that no patients are in the system.

**2.5. Probability that a patient has to wait (Erlang C)**
$${P(W_q>0)=\frac{(\lambda/\mu)^c}{c!(1-\rho)}P_0}$$
This formula estimates the probability that all doctors are busy when a new patient arrives.

**2.6. Average waiting time in the queue $(W_q)$**

The expected waiting time before a patient starts their consultation is:
$${W_q=\frac{P(W_q>0)}{c\mu-\lambda}}$$

**2.7. Average total time in the hospital service unit $(W)$**

This includes both waiting time and examination time:
$${W=W_q+\frac{1}{\mu}}$$

**2.9. Average number of patients waiting $(L_q)$**

Using Little's Law:
$${L_q=\lambda W_q}$$

This represents the average number of patients waiting for a doctor.

## 3. Optimization model
The main objective is to be minimized the average patient waiting time by choosing the appropriate number of doctors.

A simple optimization objective is:
${\min_{c} W_q(c)}$

Subject to:
$c\mu>\lambda$, $c\in\mathbb{Z}^{+}$

However, minimizing waiting time alone would always favor having more doctors. Therefore, a more realistic objective includes staffing costs:
${\min_c Z(c)=W_q(c)+\gamma C(c)}$

Where:
- $W_q(c)$ = average waiting time in minutes.
- $C(c)=cC_d$ = total staffing cost per hour.
- $C_d$ = hourly cost of one doctor.
- $\gamma$ = coefficient balancing waiting time and cost.
We can also introduce a service-level requirement:
$W_q(c)\leq W_{\max}$

For example, the hospital may require the average waiting time to remain below 15 minutes.

The optimal number of doctors is the smallest-cost staffing configuration that meets the waiting-time target, or the configuration that minimizes the combined objective.

## 4. Complete Python implementation

Hospital data:

| Parameter | Value | Explanation |
|---|---|---|
| $\lambda$ | 24 patients/hour | Average patient arrivals |
| $\mu$ | 4 patients/hour | Average patients examined by one doctor |
| $c$ | 1-10 | Number of doctors to evaluate |
| $W\text{max}$ | 15 minutes | Maximum acceptable average waiting time |
| $C_d$ | 30 units/hour | Cost of one doctor per hour |


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math

# Hospital parameters
arrival_rate = 24          # patients per hour
service_rate = 4           # patients per doctor per hour

max_doctors = 10
waiting_time_target = 15   # minutes

doctor_cost = 30           # cost units per doctor per hour

# Number of doctors to evaluate
doctors_range = range(1, max_doctors + 1)
```

### 4.1. Implementation of the M/M/c mathematical model
This function calculates the queueing performance for a given number of doctors.

```python
def mmc_model(lam, mu, c):

    # Stability condition
    if c * mu <= lam:
        return {
            "doctors": c,
            "utilization": np.nan,
            "probability_wait": np.nan,
            "waiting_time": np.inf,
            "total_time": np.inf,
            "queue_length": np.inf,
            "cost": c * doctor_cost
        }

    # Traffic intensity
    rho = lam / (c * mu)

    # Offered load
    a = lam / mu

    # Probability of an empty system
    sum_terms = sum(
        (a ** n) / math.factorial(n)
        for n in range(c)
    )

    last_term = (
        (a ** c) /
        (math.factorial(c) * (1 - rho))
    )

    p0 = 1 / (sum_terms + last_term)

    # Erlang C: probability that a patient must wait
    probability_wait = last_term * p0

    # Average waiting time in hours
    waiting_time = probability_wait / (c * mu - lam)

    # Convert waiting time into minutes
    waiting_time_minutes = waiting_time * 60

    # Average total time in the system
    total_time = waiting_time + (1 / mu)

    # Average number of patients waiting
    queue_length = lam * waiting_time

    return {
        "doctors": c,
        "utilization": rho * 100,
        "probability_wait": probability_wait * 100,
        "waiting_time": waiting_time_minutes,
        "total_time": total_time * 60,
        "queue_length": queue_length,
        "cost": c * doctor_cost
    }
```

### 4.2. Evaluation of different staffing configurations
Calculation of the expected waiting time for 1, 2, 3, and up to 10 doctors:

```python
results = []

for c in doctors_range:

    result = mmc_model(
        arrival_rate,
        service_rate,
        c
    )

    results.append(result)

df = pd.DataFrame(results)

print(df.round(2))
```

### 4.3. Identification of the optimal number of doctors
The configuration with the lowest staffing cost that meets the hospital's average waiting-time target would be selected:

```python
feasible = df[
    (df["waiting_time"] <= waiting_time_target)
]

if len(feasible) > 0:

    optimal = feasible.sort_values(
        "cost"
    ).iloc[0]

    optimal_doctors = int(optimal["doctors"])

    print("Optimal number of doctors:", optimal_doctors)

    print(
        "Average waiting time:",
        round(optimal["waiting_time"], 2),
        "minutes"
    )

    print(
        "Doctor utilization:",
        round(optimal["utilization"], 2),
        "%"
    )

    print(
        "Staffing cost per hour:",
        optimal["cost"]
    )

else:
    print("No tested staffing configuration meets the waiting-time target.")
```

## Conclusion

Queueing theory offers a shared quantitative language for patient flow, citizen-facing administration, and passenger boarding. Analytical M/M/c models are useful for quick capacity and staffing calculations in pooled service systems. Discrete-event simulation is more flexible when the process includes finite capacity, batch service, priorities, multiple stages, or operational rules.

The central policy lesson is to balance capacity, waiting time, cost, service quality, and equitable access. A model should support—not replace—professional judgment and public accountability.
