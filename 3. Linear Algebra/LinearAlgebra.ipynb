## Functions

```python
x = np.linspace(start = -1, stop = 1, num = 10000)
```

```python
y1 = np.sqrt(1 - x ** 2)
y2 = -np.sqrt(1 - x ** 2)
```

```python
plt.plot(x, y1, c = "red")
plt.plot(x, y2, c = "red")

plt.axis("equal")
plt.show()
```

```python
phi = np.linspace(start = 0, stop = 2 * np.pi, num = 10000)
```

```python
r= [1] * len(phi)
```

```python
plt.polar(phi, r)
plt.show()
```

```python
phi = np.linspace(start = 0, stop = 2 * np.pi, num = 10000)
r = 1
x_circle = r * np.cos(phi)
y_circle = r * np.sin(phi)

plt.plot(x_circle, y_circle)
plt.gca().set_aspect("equal")
plt.show()
```

```python
phi = np.linspace(start = 0, stop = 2 * np.pi, num = 10000)
r = np.sin(2 * phi)
x_test = r * np.cos(phi)
y_test = r * np.sin(phi)

plt.plot(x_test, y_test)
plt.gca().set_aspect("equal")
plt.show()
```
