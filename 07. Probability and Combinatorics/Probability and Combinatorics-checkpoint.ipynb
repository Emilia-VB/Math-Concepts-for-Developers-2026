```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
```

# Probability and Combinatorics
## LAB

```python
results = np.random.randint(1, 7, 1000)
```

```python
plt.plot(results)
plt.xlabel("Experiment number")
plt.ylabel("Outcome")
plt.show()
```

```python
def throw_dice():
    return np.random.randint(1, 7) 
x = [throw_dice() for i in range(100)]
plt.plot(x)
plt.show()
```

```python
np.sum(results == 1), np.sum(results == 2), np.sum(results == 3)
```

```python
plt.hist(results)
plt.show()
```

```python
counts = Counter(results)
```

```python
plt.bar(counts.keys(), counts.values())
plt.xlabel("Outcome")
plt.ylabel("Occurences")
plt.show()
```
