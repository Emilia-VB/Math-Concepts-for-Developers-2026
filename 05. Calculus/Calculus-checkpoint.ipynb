## Limits

```python
def f(x):
    return x ** 3 - 2 * x + 5
```

```python
f(1)
```

```python
f(2)
```

```python
x_central = 1
```

```python
delta = 0.1
x_min, x_max = x_central - delta, x_central + delta
y_min, y_max = f(x_min), f(x_max)
print(f"If x: [{x_min}; {x_max}],")
print(f"then y: [{y_min}; {y_max}]")
```

```python
delta = 0.00001
x_min, x_max = x_central - delta, x_central + delta
y_min, y_max = f(x_min), f(x_max)
print(f"If x: [{x_min}; {x_max}],")
print(f"then y: [{y_min}; {y_max}]")
```

```python
def get_limit(f, a):
    epsilon = np.array([10 ** p for p in np.arange(0, -11, -1, dtype = float)])
    x = np.append(a - epsilon, (a + epsilon)[::-1])
    y = f(x)
    return np.stack([x, y], axis = 1)
print(get_limit(lambda x: x ** 2, 3))
print(get_limit(lambda x: x ** 2 + 3 * x, 2))
print(get_limit(lambda x: np.sin(x), 0))
```

## Derivatives

```python
def calculate_derivative(f, x_0):
    h = 0.0001
    return (f(x_0 + h) - f(x_0)) / h
calculate_derivative(lambda x: 3 * x**2 + 5 * x - 8, 3)
```

```python
def calculate_derivative(f, x_0, tolerance = 0.0000001):
    return (f(x_0 + tolerance) - f(x_0)) / tolerance
calculate_derivative(lambda x: 3 * x**2 + 5 * x - 8, 3)
```

```python
calculate_derivative(np.sin, 0)
```

## Integrals

```python
def integral(f, x_min, x_max, num_points = 100000):
    dx = (x_max - x_min) / num_points
    x = np.linspace(x_min, x_max, 100000)
    y = f(x)
    result = np.sum(y) * dx
    return result
```

## Calculus in many dimensions
### Gradient descent

```python
x_old = 0
x_new = 6
step_size = 0.01
precision = 0.00001
def df(x):
    # f'(x^4 - 3x^3 + 2) = 4x^3 - 9x^2
    y = 4 * x ** 3 - 9 * x ** 2
    return y
while abs(x_new - x_old) > precision:
    x_old = x_new
    x_new += -step_size * df(x_old)
print("The local minimum occurs at ", x_new)
```
